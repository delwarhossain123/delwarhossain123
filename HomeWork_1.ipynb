#Problem 3: Student Grade Calculator

```python
Marks=int(input("Enter Marks :"))


if (100 >= Marks >= 90) :
  print("You obtained : A+")
elif(89<= Marks >80) :
  print("You obtained : A")
elif(79 >= Marks >=70 ) :
  print("You obtained : B")
elif(69 >= Marks > 60):
  print("You obtained : C")
```

#Problem 4: Bus Ticket Discount
