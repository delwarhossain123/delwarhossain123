#Problem 8: Leap Year Checker

```python
Year=int(input("Enter year :"))
if Year%400 == 0:
  print("Leap year")
elif Year%4 == 0  and Year%100 != 0 :
  print(" Leap year")
else :
  print("Not Leap year")

```
